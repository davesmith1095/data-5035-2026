```python
rates = json.loads(negotiated_rates)
# JSON dictionary/list order is DLDLDDDLDL
```

### Gameplan (also, AI Prompt)
* For each Reporting Entity Name, get name
  * From that reporting entity's out of network list
    * Get billing code type, billing code, and description
    * Get the allowed amount
      * From that allowed amount list
        * Get service_code and billing_class
        * From the allowed amount's providers list

* Only write full rows (no nulls for more detailed rows) so we have...

```python
# Rewrote AI code to think through process, added comments, made adjustments

# Start with a blank list
data = []
# From that reporting entity's out of network LIST
for out_of_network in rates['out_of_network']:
    # From that allowed amount LIST
    for allowed_amounts in out_of_network['allowed_amounts']:
        # From the allowed amount's payments DICT
        payments_dict = allowed_amounts['payments']
            # From payments --> providers LIST
        for provider in payments_dict['providers']:
            #Create row dict to hold key (column) value (row) pairs
            row = {
                # Display in logical order of "levels" for readability
                'name': out_of_network['name'],
                'billing_code_type': out_of_network['billing_code_type'],
                'billing_code': out_of_network['billing_code'],
                'description': out_of_network['description'],
                'service_code': allowed_amounts['service_code'],
                'billing_class': allowed_amounts['billing_class'],
                'allowed_amount': payments_dict['allowed_amount'],
                'billed_charge': provider['billed_charge'],
                'npi': provider['npi'][0]
            }
            data.append(row)

df = spark.createDataFrame(data)

df = df.select(
    'name', 'billing_code_type', 'billing_code', 'description',
    'service_code', 'billing_class', 'allowed_amount', 'billed_charge', 'npi'
)

display(df)
```

```python
# Trying Spark Explode approach, mimics Looping with Explodes

# Create dataframe from negotiated_rates passed in as a list of tuples
df_raw = spark.createDataFrame([(negotiated_rates,)], ["json_string"])

# Use spark to detect schema
schema = spark.range(1) \
    .select(schema_of_json(lit(negotiated_rates))) \
    .collect()[0][0]

# Parse the string into sdf dataframe using detected schema
sdf = df_raw.select(from_json(col("json_string"), schema).alias("data")) \
            .select("data.*") # Unpack the 'data' struct to get the original columns

# Use explode on lists similar to steps in Python
explode1 = sdf.withColumn("oon", explode("out_of_network"))

# Iteratively add new columns with exploded data
explode2 = explode1.withColumn("aa", explode("oon.allowed_amounts"))
explode3 = explode2.withColumn("pr", explode("aa.payments.providers"))

# Select final columns (can use dot notation to "traverse" dictionaries, e.g., allowed_amount)
finaldf = explode3.select(
    col("oon.name"),
    col("oon.billing_code"),
    col("oon.billing_code_type"),
    col("oon.description"),
    col("aa.service_code"),
    col("aa.billing_class"),
    col("aa.payments.allowed_amount"),
    col("pr.billed_charge"),
    col("pr.npi")[0].alias("npi")
)

display(finaldf)
```

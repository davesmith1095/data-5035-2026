```python
# Read the CSV
df = {
    spark.read
        .option("header", "true")
        .option("inferSchema", "true")
        # lazy execution reviews file before acting on it (e.g. output)
        .csv("hospitals.csv")
}

# Aggregate by city
result = {
    df.groupby("city")
        .agg(
            F.count("*").alias("num_hospitals"),
            F.sum("beds").alias("total_beds")
        )
}

# Show results (Collect is also an option)
result.show()
```

```
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://happenings.washu.edu/calendar/day/2026/4/27"
response = requests.get(url)
soup = BeautifulSoup(response.content, 'html.parser')

events = []
for card in soup.find_all('div', class_='em-card_text'):
    name = card.find('h3', class_='em-card_title').get_text(strip=True) if card.find('h3', class_='em-card_title') else None
    
    date_elem = card.find('p', class_='em-card_event-text')
    date = date_elem.get_text(strip=True) if date_elem else None
    
    location_elem = card.find('i', class_='fa-map-marker-alt')
    location = location_elem.parent.get_text(strip=True) if location_elem else None
    
    events.append({'Name': name, 'Date': date, 'Location': location})

df = pd.DataFrame(events)
df
```

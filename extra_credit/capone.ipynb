```python
import pandas as pd
```

```python
# ingest the data, transform to dataframe
df = pd.read_csv('capone.csv')
```

```python
# udpates to dataframe for analysis

# Separate out letters/numbers from the door code
df['letter'] = [''.join(l for l in x if l.isalpha()) for x in df['number']]
df['dc_number'] = [''.join(n for n in x if n.isnumeric()) for x in df['number']]

# Use ord method to find the order/numerical value of the alpha char
df['letter_ord'] = df['letter'].apply(lambda x: ord(x) - 64 if x != '' else None)   

# Rename then reorder df columns for clarity
df = df.rename(columns={'number':'door_code', 'letter': 'dc_letter'})
df = df[['door_code', 'dc_number', 'dc_letter', 'letter_ord', 'response_code']]

df
```
